```python
import os
import datasets

evaluation_set = datasets.load_from_disk("./kasra_run_evlaution_on_this_original")
evaluation_set
```

```python
query = evaluation_set['query']
instruction = "You are a professional requirements engineer who helps users brainstorm more software requirements."
```

```python
from openai import OpenAI

base_url = "https://api.deepseek.com"
api_key = "blinded-for-privacy"

client = OpenAI(api_key = api_key, base_url = base_url)
```

```python
def run_deepseek_model(model_name, query, instruction):
    response = client.chat.completions.create(
        model=model_name,
        messages=[
            {"role": "system", "content": instruction},
            {"role": "user", "content": query},
        ],
    )
    return response.choices[0].message.content
```

```python
model_name = "deepseek-chat"
deepseek_chat_v_32_25_12_01 = []

for q in query:
    deepseek_response = run_deepseek_model(model_name, q, instruction)
    deepseek_chat_v_32_25_12_01.append(deepseek_response)
```

```python
evaluation_set = evaluation_set.add_column(f"deepseek_chat_v_32_25_12_01", deepseek_chat_v_32_25_12_01)
```
