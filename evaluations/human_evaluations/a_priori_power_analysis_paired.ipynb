# A Priori Power Analysis for the Paired Evaluation Design

This notebook uses the same parameters as the original analysis and replaces the independent-samples analogues with paired-samples analogues. Sample sizes are reported as matched pairs, not as the combined number of requirements.

```python
# Common imports

import math
from scipy.stats import norm
from statsmodels.stats.power import TTestPower
```

```python
# Parameters for the power analysis

effect_size = 0.5      # Moderate standardized effect
alpha = 0.05           # Significance level
power_desired = 0.8    # Desired power (80%)
```

## Paired Wilcoxon signed-rank tests

A paired t-test, implemented as a one-sample t-test on within-pair differences, is used as the parametric power-analysis analogue of the Wilcoxon signed-rank test.

```python
paired_analysis = TTestPower()

sample_size_1sided = paired_analysis.solve_power(effect_size = effect_size, alpha = alpha, power = power_desired, alternative = 'larger')
print('H3-H4 required matched pairs (one-sided):', math.ceil(sample_size_1sided))

sample_size_2sided = paired_analysis.solve_power(effect_size = effect_size, alpha = alpha, power = power_desired, alternative = 'two-sided')
print('H5-H6 required matched pairs (two-sided):', math.ceil(sample_size_2sided))
```

## Paired binary tests (McNemar)

The asymptotic paired-proportion test is used as the power-analysis analogue of the exact McNemar test. Here, the numerical effect-size value of 0.5 is interpreted as a standardized paired-proportion effect rather than Cohen's d.

```python
def paired_binary_sample_size(effect_size, alpha = 0.05, power = 0.8, alternative = 'two-sided'):
    if alternative == 'larger': z_alpha = norm.ppf(1 - alpha)
    elif alternative == 'two-sided': z_alpha = norm.ppf(1 - alpha / 2)
    else: raise ValueError("alternative must be 'larger' or 'two-sided'")
    z_power = norm.ppf(power)
    return math.ceil(((z_alpha + z_power) / effect_size) ** 2)
```

```python
h1_sample_size = paired_binary_sample_size(effect_size, alpha, power_desired, alternative = 'larger')
print('H1 required matched pairs (one-sided):', h1_sample_size)

h2_sample_size = paired_binary_sample_size(effect_size, alpha, power_desired, alternative = 'two-sided')
print('H2 required matched pairs (two-sided):', h2_sample_size)
```

## One-sample Wilcoxon signed-rank tests

A one-sided one-sample t-test is retained as the analogue for H7-H9. The result is rounded upward with `math.ceil`.

```python
one_sample_analysis = TTestPower()

one_sample_size = one_sample_analysis.solve_power(effect_size = effect_size, alpha = alpha, power = power_desired, alternative = 'larger')
print('H7-H9 required independent requirements (one-sided):', math.ceil(one_sample_size))
```
